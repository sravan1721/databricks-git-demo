```python
from pyspark.sql import  functions as F
data = [
    (1,"Miller",6000),
    (2,"Blake",9000),
    (3,"Kiran",7000)
]
columns=["eid","ename","sal"]
df=spark.createDataFrame(data,columns)
df.show()
df=df.withColumn("sal_cat",F.when(df.sal>8000,"High").when(df.sal>5000,"Medium").otherwise("Low"))
df.show()
```
